```python
import glob

import pandas as pd
import ampworks as amp

base = amp.read_table('base.txt')
base['DateTime'] = pd.to_datetime(base['DateTime'])

for f in glob.glob('*.txt'):
    ds = amp.read_table(f)

    ds['Seconds'] = ds['Seconds'].round(1)
    ds['DateTime'] = pd.to_datetime(ds['DateTime'])

    print(f)
    pd.testing.assert_frame_equal(
        base.sort_index(axis=1),
        ds.sort_index(axis=1),
    )
```

```python
arbin = amp.read_excel('arbin1.xlsx', sheet_name=4)

arbin['Seconds'] = arbin['Seconds'].round(1)
arbin['DateTime'] = pd.to_datetime(arbin['DateTime'])

pd.testing.assert_frame_equal(
    base.sort_index(axis=1),
    arbin.sort_index(axis=1),
)
```

```python
bitrode = amp.read_csv('bitrode1.csv')

bitrode['Seconds'] = bitrode['Seconds'].round(1)
bitrode['DateTime'] = pd.to_datetime(bitrode['DateTime'])

pd.testing.assert_frame_equal(
    base.sort_index(axis=1),
    bitrode.sort_index(axis=1),
)
```
